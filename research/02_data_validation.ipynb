# Stage 02 — Data Validation

This notebook loads the dataset and schema through the production configuration manager, validates columns, types, missing values, and the target column, then inspects the validation result. Run Stage 01 first.

```python
import os
from pathlib import Path

if Path.cwd().name == "research":
    os.chdir("..")
PROJECT_ROOT = Path.cwd()
PROJECT_ROOT
```

```python
import pandas as pd

from mlproject.components.data_validation import DataValidation
from mlproject.config.configuration import ConfigurationManager
```

```python
configuration = ConfigurationManager()
validation_config = configuration.get_data_validation_config()
validation_config
```

```python
data = pd.read_csv(validation_config.unzip_data_dir)
schema_comparison = pd.DataFrame({
    "expected_dtype": validation_config.all_schema,
    "actual_dtype": {column: str(data[column].dtype) for column in data.columns},
})
print(f"Dataset shape: {data.shape}")
print(f"Missing values: {int(data.isna().sum().sum())}")
schema_comparison
```

```python
validator = DataValidation(config=validation_config)
validation_passed = validator.validate_all_columns()
print(validation_config.status_file.read_text(encoding="utf-8").strip())
validation_passed
```
