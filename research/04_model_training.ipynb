# Stage 04 — Model Training

This notebook productionizes the estimator choices explored in `Expriement.ipynb`. It compares scaled `LogisticRegressionCV` with the existing `ElasticNet` experiment using weighted multiclass metrics, selects the best validation F1 score, retrains on the full training set, and saves `model.pkl`.

```python
import os
from pathlib import Path

if Path.cwd().name == "research":
    os.chdir("..")
PROJECT_ROOT = Path.cwd()
PROJECT_ROOT
```

```python
import joblib
import pandas as pd

from mlproject.components.model_trainer import ModelTrainer
from mlproject.config.configuration import ConfigurationManager
```

```python
configuration = ConfigurationManager()
trainer_config = configuration.get_model_trainer_config()
training_data = pd.read_csv(trainer_config.train_data_path)
print(f"Training shape: {training_data.shape}")
print(f"Target: {trainer_config.target_column}")
print(f"Validation size: {trainer_config.validation_size}")
print(f"Random state: {trainer_config.random_state}")
trainer_config
```

```python
feature_columns = training_data.drop(columns=[trainer_config.target_column]).columns.tolist()
target_distribution = training_data[trainer_config.target_column].value_counts().sort_index()
print(f"Feature count: {len(feature_columns)}")
print(feature_columns)
target_distribution
```

```python
trainer = ModelTrainer(config=trainer_config)
selected_model_name = trainer.train()
print(f"Selected model: {selected_model_name}")
```

```python
model_path = trainer_config.root_dir / trainer_config.model_name
trained_model = joblib.load(model_path)
print(f"Saved model: {model_path}")
print(f"Model type: {type(trained_model).__name__}")
print(f"Feature names: {list(trained_model.feature_names_in_)}")
getattr(trained_model, "named_steps", trained_model)
```
