# Stage 03 — Data Transformation

This notebook uses the production transformation component and the split settings from `params.yaml`. The split is stratified by `quality` so each quality class remains represented. Run Stages 01 and 02 first.

```python
import os
from pathlib import Path

if Path.cwd().name == "research":
    os.chdir("..")
PROJECT_ROOT = Path.cwd()
PROJECT_ROOT
```

```python
import pandas as pd

from mlproject.components.data_transformation import DataTransformation
from mlproject.config.configuration import ConfigurationManager
```

```python
configuration = ConfigurationManager()
transformation_config = configuration.get_data_transformation_config()
transformation_config
```

```python
transformer = DataTransformation(config=transformation_config)
transformer.train_test_splitting()
```

```python
train_data = pd.read_csv(transformation_config.root_dir / "train.csv")
test_data = pd.read_csv(transformation_config.root_dir / "test.csv")
print(f"Training shape: {train_data.shape}")
print(f"Test shape: {test_data.shape}")

class_distribution = pd.DataFrame({
    "train_fraction": train_data[transformation_config.target_column].value_counts(normalize=True).sort_index(),
    "test_fraction": test_data[transformation_config.target_column].value_counts(normalize=True).sort_index(),
})
class_distribution
```
